```python
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
# veri olustur
X = np.random.rand(100,1) 
y = 3 + 4 * X + np.random.rand(100,1)          # y = b + ax

plt.scatter(X, y)
```

```python
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

```python
plt.figure()
plt.scatter(X,y)
plt.plot(X, lin_reg.predict(X), color="red", alpha=0.7)
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear Regression")
```

```python
# katsayi ve sabit bayes degeri

a = lin_reg.coef_[0][0]
display(a)        # a

b = lin_reg.intercept_[0]   # b
display(b)

print(f"b+xa => {b}+x{a}")
```

```python
for i in range(100):
    y_ = b + X * a + np.random.rand(100,1)
    plt.plot(X, y_, color="black", alpha=0.2)
    plt.scatter(X, y_, color="green", alpha=0.2)
```
